```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

```

```python
#calculate total transaction value without using sum().

Total = 0
for i in transactions:
  Total = Total + i
print("Sum of total transcation value is :",Total)
```

```python
#From the same list, find the highest and lowest transaction without max() and min().

highest = transactions[0]
lowest = transactions[0]
for i in transactions:
  if i > highest:
    highest = i

  elif i < lowest:
    lowest = i

print("Highest transaction value is :",highest)
print("Lowest transaction value is :",lowest)
```

```python
marks = [78, 92, 45, 67, 88, 53, 99]
```

```python
#count how many students scored
stud = 0
avg_stud = 0
stud_three = 0
below = 0
for i in marks:
  if i > 90:
    stud = stud + 1
  elif i > 75 and i < 89:
    avg_stud = avg_stud + 1
  elif i > 50 and i < 74:
    stud_three = stud_three + 1
  else:
    below = below + 1
print("Number of students scored 90+ is :",stud)
print("Number of students scored 75-89 is :",avg_stud)
print("Number of students scored 50-74 is :",stud_three)
print("Number of students scored below 50 is :",below)

```

```python
correct_password = "mypass12345"
max_attempts = 3
attempts = 0

while attempts < max_attempts:
    password = input(str("Enter your password :"))
    if password == correct_password:
        print("Login successful!")
        break
    else:
        attempts = attempts + 1
        if attempts < max_attempts:
            print(f"Incorrect password. You have {max_attempts - attempts} attempts remaining.")
        else:
            print("You have exceeded the maximum number of attempts. Account locked.")
```

```python
#Count the frequency of every character in a string without using Counter.

string = input("Enter a string: ")
frequency = {}
for i in string:
  if i in frequency:
    frequency[i] = frequency[i] + 1
  else:
    frequency[i] = 1
print("Frequency of characters in the string is:", frequency)
```

```python


l = [10, 20, 30, 40, 50]
```

```python
#Find the second-largest number in a list without using sort().
maax = l[0]
second_max = l[0]
for i in l:
  if i > maax:
    maax = i
for j in l:
  if j > second_max and j < maax:
      second_max = j
print("The Second largest number in the list is:", second_max)
```

```python
#Check whether a string is a palindrome using loops.

string = input("Enter a string: ")
reverse = ""
for i in string:
  reverse = i + reverse
if string == reverse:
  print("The string is a palindrome")
else:
  print("The string is not a palindrome")
```

```python
balance = 10000  # Initial balance

while True:
    print("\n--- ATM Menu ---")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        print(f"Your current balance is: ₹{balance}")
    elif choice == '2':
        try:
            deposit_amount = float(input("Enter amount to deposit: ₹"))
            if deposit_amount > 0:
                balance += deposit_amount
                print(f"₹{deposit_amount} deposited successfully. New balance: ₹{balance}")

            else:
                print("Deposit amount must be positive.")
        except ValueError:
            print("Invalid input. Please enter a number.")
    elif choice == '3':
        try:
            withdraw_amount = float(input("Enter amount to withdraw: ₹"))
            if withdraw_amount > 0:
                if balance >= withdraw_amount:
                    balance -= withdraw_amount
                    print(f"₹{withdraw_amount} withdrawn successfully. New balance: ₹{balance}")

                else:
                    print("Insufficient balance.")
            else:
                print("Withdrawal amount must be positive.")
        except ValueError:
            print("Invalid input. Please enter a number.")
    elif choice == '4':
        print("Thank you for using the ATM. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 4.")
```
