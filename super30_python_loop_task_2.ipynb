```python
#Print numbers from 1–100 but skip numbers divisible by 5 using continue.
for i in range(1,100):
    if i%5==0:
        continue
    print(i)
```

```python
#Iterate from 1–100 and stop when you encounter the first number divisible by both 7 and 11.

for i in range(1,100):
  if i%7==0 and i%11==0:
    break

  print(i)

```

```python
#Search for a user-provided number inside a list. Use for-else to print:
#  "Number Found" , "Number Not Found"


l = [44,32,112,54,68,98,10]
number = int(input("Enter a number: "))
for i in l:
  if i == number:
    print("Number Found")
    break
else:
  print("Number Not Found")
```

```python
#Generate multiplication tables from 1 to 10 using nested loops.

for i in range(1, 11):
  print(f"Multiplication table for {i}:")
  for j in range(1, 11):
    print(f"{i} x {j} = {i * j}")
  print()
```

```python
#Find all numbers between 1 and 200 divisible by both 3 and 5.

for i in range(1, 201):
  if i % 3 == 0 and i % 5 == 0:
    print(i)
```

```python
# Given a list containing duplicate elements, create another list containing only unique elements without using set().

l = [11, 52, 22, 33, 4, 45, 5, 67, 6, 67,11,52]
unique_list = []
for i in l:
  if i not in unique_list:
    unique_list.append(i)
print(unique_list)
```

```python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]
```

```python
# count - positive numbers , negative numbers, zeros

positive_count = 0
negative_count = 0
zero_count =0

for i in numbers:
  if i > 0:
    positive_count += 1
  elif i < 0:
    negative_count += 1
  else:
    zero_count += 1

print("Positive count:", positive_count)
print("Negative count:", negative_count)
print("Zero count:", zero_count)
```

```python
#Write a program to determine whether a number is prime using a loop.

num = int(input("Enter a number: "))
for i in range(2, num):
  if num % i == 0:
    print(num, "is not a prime number")
    break
else:
  print(num, "is a prime number")
```

```python
#Print all prime numbers between 1 and 100.

for num in range (2,101):
  for i in range (2, num):
    if num % i == 0:
      break
  else:
    print(num)
```
