```python
#Print all even numbers from 1 to 100.

for i in range (1,100):
  if i%2==0:
    print(i)
```

```python
#Print all odd numbers from 1 to 100.
for i in range (1,100):
  if i%2==0:
    continue
  print(i)

```

```python
#Take an integer n and print its multiplication table from 1 to 20.

n = int(input("Enter a number: "))

for i in range(1, 21):
    print(n, "x", i, "=", n*i)
```

```python
#Calculate the sum of numbers from 1 to n using a loop.
n = int(input("Enter a number: "))
total = 0
for i in range (1,n):
  total = total + i
print("Sum of the numbers :", total)

```

```python
#Calculate the factorial of a number without using any built-in factorial function.
n = int(input("Enter a number: "))
total = 1

for i in range (1,n):
    total = total * i
print("Factorial :", total)


```

```python
#Count how many vowels exist in a user-provided string.

word = input(str("enter a word : "))
count = 0
for i in word:
  if i in "aeiouAEIOU":
    count = count + 1
print(count)
```

```python
#Reverse a string using a for loop without using [::-1] or reversed().
word = input(str("enter a word : "))
reversed_word = ""
for i in range(len(word) - 1, -1, -1):
    reversed_word += word[i]
print(reversed_word)
```
