```python
#Create a function using *args that returns the largest supplied number.

def largnum(*args):
    largnum = args[0]
    for i in args:
        if i > largnum:
            largnum = i
    return ("Largest number of given: ", largnum)
```

```python
largnum(90,989,889,212,883,172,237)
```

```python
#Write a recursive function to calculate factorial.

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
#Write a recursive function to calculate 1 + 2 + 3 + ... + n

def addnum(num):
  if num == 0:
    return 0
  else:
    return num + addnum(num-1)

```

```python
addnum(31)
```

```python
#Write a recursive function to generate the Fibonacci sequence or calculate the nth Fibonacci number.

def cal_fib(w):
  if w <= 0:
    return 0
  elif w == 1:
    return 1
  else:
    return cal_fib(w-1) + cal_fib(w-2)
```

```python
def generate_fibonacci_sequence(n):
  fib_sequence = []
  for i in range(n):
    fib_sequence.append(cal_fib(i))
  return fib_sequence

# Generate the first 12 Fibonacci numbers
print(generate_fibonacci_sequence(11))
```
