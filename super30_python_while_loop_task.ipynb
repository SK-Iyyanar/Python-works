```python
#Calculate the sum of digits of a number.

num = int(input("enter a number"))
total = 0

while num>0:
  digit = num %10
  total = total + digit
  num = num//10
print("the sum of digits is",total)
```

```python
#Reverse an integer using a while loop.

num = int(input("enter a number"))
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reverse number:", rev)
```

```python
#Calculate factorial using while.
num = int(input("enter a number :"))
Factorial = 1

while num > 0:
    Factorial = num * Factorial
    num = num -1

print("Factorial number:", Factorial)
```

```python
#Create a program that repeatedly asks for numbers and stops only when the user enters 0. Display the sum of all previously entered numbers.
num = int(input("enter a number : "))
total = 0

while num != 0:
  total = total + num
  num = int(input("enter a number : "))
print("Total of entered number:", total)
```

```python

options = {1: "Add", 2: "Subtract", 3: "Multiply", 4: "Divide", 5: "Exit"}
print(options)

num = int(input("Choose a number between 1 to 5 to proceed -- "))
while num in options:
  print(f" you have selected option {num, options[num]}")
  if num == 5:
    break
  num = int(input("Choose a number between 1 to 5 to proceed -- "))
```
