```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
print(numbers)
```

```python
## converting List to set and printing the data types and complete set.  

sa = set(numbers)
print(type(sa))
print(sa)
```

When converting the List varaibles numbers to set . The duplicates number in the list is removed as set data types will not take duplicate values. Duplicate values are (10, 20, 30). These duplicates are removed and got set data as above. 

```python
## converting set back to List 

new_numb = list(sa)
print(type(new_numb))
print(new_numb)
```

```python
#Print the number of original elements

print(numbers , f"length: {len(numbers)}")
```

```python
# Print the number of unique elements

print(new_numb , f"length: {len(new_numb)}")
```
