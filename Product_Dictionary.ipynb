```python
laptop = {

"brand": "Dell",

"model": "XPS 15",

"price": 120000,

"ram": "16GB",

"storage": "512GB SSD",

"available": True

}
```

```python
#Display the product

print(laptop)
```

```python
# Print brand

# Print model

# Print price

print(laptop["brand"])
print(laptop["model"])
print(laptop["price"])
```

```python
#Change price

laptop["price"] = 110000
```

```python
print(laptop["price"])
```

```python
# Add processor information

# Add GPU information

laptop.update({"processor_information": "Intel Core i7"})
laptop.update({"GPU_information": "NVIDIA GeForce RTX 3060"})
```

```python
print(laptop)
```

```python
#Change RAM to "32GB"

laptop["ram"] = "32GB"
```

```python
print(laptop["ram"])
```

```python
#Remove "available"

laptop.pop("available")
```

```python
print(laptop)
```

```python
#Print all keys


print(laptop.keys())
```

```python
#Print all values


print(laptop.values())
```

```python
#Print all items

print(laptop.items())
```
