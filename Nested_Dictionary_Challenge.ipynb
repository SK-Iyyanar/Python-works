Explain how nested dictionaries are accessed.

Nested dictionaries is a dictionary that contains another dictionaries inside it.

To access the values in dictionary we call the variable name and key name, unlike other data type we use index. But in dictionary we use key name  
ex : employee['name']

To access the values in nested dictionaries, need to call the nested dictionary first and then key value inside the nested dictionary.
ex : employee["skills"]["IAC"]
