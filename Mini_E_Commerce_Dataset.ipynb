```python
products = [{"name": "Laptop","price": 70000,"brand": "Dell"},{"name": "Phone","price": 40000,"brand": "Samsung"},{"name": "Tablet","price": 30000,"brand": "Apple"}]
```

```python
#Print all products

print(products)
```

```python
#Print first product

print(products[0])
```

```python
#Print second product's price

print(products[1]["price"])
```

```python
#Print third product's brand

print(products[2]["brand"])
```

```python
#Change first product's price

products[0]["price"] = 80000
print(products[0])
```

```python
#Add "rating" to the second product
products[1]["rating"] = 4.5
print(products[1])
```

```python
#Add another product manually

products.append({"name": "Monitor","price": 10000,"brand": "LG"})
print(products)
```

```python
#Print the final dataset
print(products)
```

**Explain the combination**

Given  Data is combination of list and Dictionary

List : The Given data is actually list and its collection of Dictionaries.

Dictionary : Inside the List each data is a Dictionary, where the Dictionary has data in Key and value format.

Indexing: To access the data inside the list , we use index option.

ex: products[0]

Keys & Values: These are type where data will be stored in Dictionary. To access the data, we call the key and will get values for the key
ex: products[0]["price"]

Combination:

The list has the collection of dictionaries and List data's will be accessed through indexs and dictionary data will be accessed by list index first and then key of the dictionary to get the value.
