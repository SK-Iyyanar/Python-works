```python
student = {

"name": "Rahul",

"age": 22,

"course": "Python",

"city": "Bangalore",

"marks": 88

}
```

```python
#Print the complete dictionary

print(student)
```

```python
#Print the student's name

print(student["name"])
```

```python
#Print their course

print(student["course"])
```

```python
#Print all keys

print(student.keys())
```

```python
#Print all values

print(student.values())
```

```python
#Print all key-value pairs

print(student.items())
```

```python
#Change marks from 88 to 92

student["marks"] = 92
print(student["marks"])
```

```python
#Add "email"

#Add "phone"

student.update({"email": "rahul@gmail.com", "phone" : 9876543210})
print(student)
```

```python
#Remove "city"

student.pop("city")
```

```python
print(student)
```

```python
#Use get() to retrieve "name"

student.get("name")
```

```python
#Create a copy of the dictionary

new_student = student.copy()
print(new_student)
print(student)
```

### Explanation of Dictionary Methods

- **`keys()`**: This will displays all the keys in the dictionary.

- **`values()`**: This will displays all the values in the dictionary.

- **`items()`**: This will displays all the key-value pairs in the dictionary.

- **`get()`**: This will display the value for the specified `key` if the `key` is available in dictionary. If the `key` is not found, it returns `None` by default.

- **`update()`**: This method updates the dictionary with key value pairs given in method . If a key is already present in dictionary, the value will be overrided in the original dictionary. If a key is not present in the original dictionary, it is added.

- **`pop()`**: This method removes the specified `key` and its value from the dictionary. If the `key` is not found, it raises a KeyError

- **`copy()`**: This method will copy of the dictionary and creates new dictionary will not affect the original dictionary
