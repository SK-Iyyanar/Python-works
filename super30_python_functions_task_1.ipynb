```python
#Create a function that calculates factorial.

def calfact (num):
  if num == 1:
    return 1
  else:
    return num * calfact(num-1)
```

```python
calfact(6)
```

```python
#Create a function that checks whether a number is prime.


def checkprim(num):

    if num < 2:
        print(num, "is not a prime number")
        return

    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
```

```python
checkprim(19)
```

```python
#Create  def calculate_discount(price, discount=10) where the default discount is 10%.

def calculate_discount(price, discount=10):
  discount_amount = float (price / discount)
  print("Total discount amount :" ,discount_amount)
  amount = price - discount_amount
  return print("Total amount after discount :" ,amount)


```

```python
calculate_discount(109)
```

```python
#Create a function that accepts a list and returns its sum without using sum().

def list_sum(numbers):
    total = 0

    for num in numbers:
        total += num

    return total
```

```python
list_sum([61,52,43,34,25])
```

```python
#Create a function that accepts a string and returns the number of vowels.


def count_vowels(text):
    if not isinstance(text, str):
        return ("Input must be a string")

    count = 0

    for char in text.lower():
        if char in "aeiou":
            count += 1
    return count
```

```python
count_vowels("345ed.iphone,samsung")
```

```python
#Create a function that accepts a string and determines whether it is a palindrome.

def check_palindrome(text):
    if not isinstance(text, str):
        return ("Input must be a string")
    elif text == text[::-1]:
      return ("Entereted text is  palindrome")
    else:
      return ("Entereted text is not palindrome")
```

```python
check_palindrome("ohiho")
```

```python
#Create a function that accepts name, age, course and returns a formatted student profile.

def student_profile(name, age, course):
  profile = {"Student_Name": name, "Age": age, "Course": course}
  return profile
```
